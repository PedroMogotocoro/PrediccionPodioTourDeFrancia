### Revisión, filtro y union.

1. Cargar los DataFrame disponibles.

2. Analizar variables y datas para tener una mejor información conjunta.

3. Construir el mejor DataFrame para fines de clasificación.

```python
import pandas as pd
print("Top 10 del Tour de France de 2006 a 2024")
df_Tour_Top10 = pd.read_csv("DataDesTop10.csv",sep=";")
df_Tour_Top10.info()
```

```python
print("Descripción de etapas de las grandes vueltas desde el comienzo hasta el 2023, incluidas las siete grandes de una semana")
df=pd.read_csv("cyclingdata.csv",sep=";")
df.info()
df.tail()
```

Dada la información del Data Frame anterior, podemos extraer datos que nos podrán ayudar a fortalecer la predicción ya que al incluir esta información, y no solo las caracteristicas de los corredores, pueden ayudar a entender como clasificar de una mejor manera.

Para ello, filtramos los datos solo para el Tour de Francia entre los años: 2006 al 2024.

```python
df['year'] = pd.to_numeric(df['year'], errors='coerce')

df_tour = df[
    (df['race'] == 'Tour de France') &
    (df['year'] > 2005) &
    (df['year'] < 2025)
]
df_tour.info()
df_tour.tail()
```

Para efectos de una mejor selección en las variables, elegimos las siguientes:

-parcours_type: Nos interesa saber, según el tipo de etapa que proporción tiene cada una en la carrera.

-distance: El total en km.

-Vertical_meters: El total, ya que nos interesa conocer el total de desnivel positivo acumulado durante la carrera.

-profile_score: El promedio, ya que nos interesa saber que tan dura fue la carrera con el perfil promedio.

-Startlist_quality: La incial, ya que por lo general la lista de corredores Top se mantiene. Aunque hay casos donde se pierden corredores por caídas, enfermedades, etc. consideramos que en general los mejores se mantienen.

-avg_Speed_winner: Se utiliza la velocidad promedio de los ganadores de etapa como una aproximación a la intensidad y características dinámicas de la carrera, ya que refleja indirectamente factores como el perfil de las etapas, condiciones climáticas y estrategias de competencia.

Todas estás variables anteriores nos ayudan a comprender las dinamicas de cada año y como la intensidad y caracteristicas de la carrera pueden ayudar a explicar el quedar dentro o fuera del podio del **TOUR DE FRANCE**


```python
df_features = df_tour.groupby('year').agg({          
    'distance': 'sum',                                # distancia total
    'vertical_meters': 'sum',                         # desnivel total
    'profile_score': 'mean',                          # perfil promedio
    'startlist_quality': 'first',                     # calidad inicial
    'avg_speed_winner': 'mean'                        # velocidad promedio
}).reset_index()
df_features.info()
df_features.tail()
```

Manejo especial de la variable: **parcours_type**

Debido a que en esta variable se están mezclando tanto la descripción de como es el recorrido y el final, crearemos una variable que aporte un tipo de información según el recorrido y otra según su final.

```python
def tipo_terreno(x):
    if 'mountain' in x.lower():
        return 'mountain'
    elif 'hill' in x.lower():
        return 'hills'
    else:
        return 'flat'

def tipo_final(x):
    if 'uphill' in x.lower():
        return 'uphill'
    else:
        return 'flat_finish'

df_tour['terrain'] = df_tour['parcours_type'].apply(tipo_terreno)
df_tour['finish'] = df_tour['parcours_type'].apply(tipo_final)
```

```python
df_terrain = df_tour.pivot_table(
    index='year',
    columns='terrain',
    values='stage',
    aggfunc='count',
    fill_value=0
)

df_terrain = df_terrain.div(df_terrain.sum(axis=1), axis=0)
df_terrain
```

```python
df_finish = df_tour.pivot_table(
    index='year',
    columns='finish',
    values='stage',
    aggfunc='count',
    fill_value=0
)

df_finish = df_finish.div(df_finish.sum(axis=1), axis=0)
df_finish
```

```python
df_parcours_final = df_terrain.merge(
    df_finish,
    left_index=True,
    right_index=True
)
df_parcours_final
```

Ahora será necesario unir el anterior data frame a las otras variables de carrera.

```python
df_features_full = df_features.merge(
    df_parcours_final,
    on='year',
    how='left'
)
df_features_full = df_features_full.rename(columns={'year': 'Year'})#Para evitar duplicado en el año.
```

Finalmente, unimos las características de carrera a el data frame del top10.

En términos técnicos, lo que se busca es una relación de muchos a uno (many-to-one): se tienen muchos corredores para un mismo año y se quiere asignar a cada uno la información única de ese año.

```python
df_final = df_Tour_Top10.merge(
    df_features_full,
    left_on='Year',
    right_on='Year',
    how='left'
)
df_final.info()
```
